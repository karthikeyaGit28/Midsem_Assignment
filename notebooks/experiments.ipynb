# LegalLens: Information Retrieval Experiments & Evaluation
This notebook runs and visualizes the benchmark comparing **TF-IDF**, **BM25**, **Dense Semantic**, and **Hybrid Retrieval**.

```python
import os
import sys
import pandas as pd
import matplotlib.pyplot as plt

sys.path.insert(0, os.path.abspath('..'))
from src.hybrid_ranker import load_hybrid_system
from src.tfidf_retriever import TFIDFRetriever

# Load models
hybrid = load_hybrid_system()
tfidf = TFIDFRetriever.load('../data/processed/tfidf_vectorizer.pkl')
print('Models loaded successfully!')
```

```python
# Test query on Hybrid Ranker
query = 'Can a tenant be evicted without proper notice under rent control?'
results = hybrid.rank_query(query, top_k=5, alpha=0.75)
print(f'Query: {query}\n')
for i, r in enumerate(results, 1):
    print(f"{i}. {r['case_name']} (Final Score: {r['final_score']:.4f})")
    print(f"   BM25 Norm: {r['bm25_norm']:.4f} | Semantic Norm: {r['semantic_norm']:.4f}")
    print(f"   Passage: \"{r['passage'][:120]}...\"\n")
```

```python
# Display Benchmark Results
df_metrics = pd.read_csv('../results/metrics.csv')
display(df_metrics)

# Display Tuning Curve
df_tuning = pd.read_csv('../results/alpha_tuning.csv')
plt.figure(figsize=(8, 4), dpi=150)
plt.plot(df_tuning['alpha'], df_tuning['nDCG@10'], marker='o', label='nDCG@10', color='#9013FE')
plt.plot(df_tuning['alpha'], df_tuning['MRR'], marker='s', label='MRR', color='#4A90E2')
plt.xlabel('Alpha Weight (BM25 vs Semantic)')
plt.ylabel('Score')
plt.title('Sensitivity Analysis of Hybrid Fusion Alpha')
plt.legend()
plt.grid(True)
plt.show()
```
