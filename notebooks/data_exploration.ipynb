# LegalLens: Data Exploration & Schema Analysis
This notebook explores the **IndicLegalQA** dataset (Indian Supreme Court Q&A benchmark).
We inspect the raw schema, group records into canonical judgment documents, and analyze corpus statistics.

```python
import json
import os
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

# Load raw dataset
raw_path = '../data/raw/IndicLegalQA_Dataset_10K_Revised.json'
if not os.path.exists(raw_path):
    raw_path = '../IndicLegalQA Dataset_10K_Revised.json'

with open(raw_path, 'r', encoding='utf-8') as f:
    records = json.load(f)

print(f'Total Q&A records: {len(records)}')
print('Sample record keys:', list(records[0].keys()))
```

```python
# Convert to DataFrame for descriptive analysis
df = pd.DataFrame(records)
print('Unique cases:', df['case_name'].nunique())
print('Unique judgment dates:', df['judgement_date'].nunique())
df['q_words'] = df['question'].apply(lambda x: len(str(x).split()))
df['a_words'] = df['answer'].apply(lambda x: len(str(x).split()))
df[['q_words', 'a_words']].describe()
```

```python
# Load processed canonical documents
with open('../data/processed/documents.json', 'r', encoding='utf-8') as f:
    docs = json.load(f)

df_docs = pd.DataFrame(docs)
print(f'Total canonical documents: {len(df_docs)}')
print('\nLegal Category Distribution:')
print(df_docs['category'].value_counts())

# Plot category distribution
plt.figure(figsize=(8, 4), dpi=150)
df_docs['category'].value_counts().plot(kind='barh', color='#3f51b5')
plt.title('Distribution of Legal Categories in LegalLens Corpus')
plt.xlabel('Number of Judgments')
plt.tight_layout()
plt.show()
```
